```python
import json
import pandas as pd
import numpy as np

languages = ['english','german','swedish','latin','spanish','chinese','norwegian_1','norwegian_2']

model_name = 'gemini25pro'

# load path_to_defs from .env
import os
from dotenv import load_dotenv
load_dotenv()
path_to_defs = os.getenv('PATH_TO_DEF_EMBS')


rows = []

for language in languages:
    with open(f"{path_to_defs}/{language}_definitions_by_{model_name}.json", "r") as f:
        definitions = json.load(f)

    # number of target words
    num_words = len(definitions)

    # number of definitions per word
    defs_per_word = [len(defs) for defs in definitions.values()]

    mean_defs_per_word = np.mean(defs_per_word)
    std_defs_per_word = np.std(defs_per_word)


    mean_std_str = f"{mean_defs_per_word:.2f}±{std_defs_per_word:.2f}"


    rows.append({
        "language": language.capitalize(),
        "num_words": num_words,
        "mean_defs_per_word": mean_std_str,
    })

# make it into a table
df = pd.DataFrame(rows)
df.index = df.language
# drop language column
df = df.drop(['language'], axis=1)

df.to_latex
```
